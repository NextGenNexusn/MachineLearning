``````python
numerical_cols=df.select_dtypes(
    include=['int64','float64']
).columns

print(numerical_cols)
``````

``````python
print(df[numerical_cols].describe())


  # This gives us eight important statistical values.
          # Statistic	Meaning    ----- count	Number of non-missing observations
          # mean	               -----Average value
          # std	                 -----Standard deviation
          # min	                 -----Minimum value
          # 25%	                 -----First quartile (Q1)
          # 50%	                 -----Median (Q2)
          # 75%	                 -----Third quartile (Q3)
          # max	                 -----Maximum value
``````

``````python
# print mean of every numerical cols
for col in numerical_cols:
  print(col ,":", df[col].mean())

# print meadian of every numerical cols
for col in numerical_cols:
  print(col,":",df[col].median())

#print mode of every numerical cols :
for col in numerical_cols:
  print(col ,":",df[col].mode().tolist())
``````

``````python
#                                   Measure of dispersion
# 1. Range
for col in numerical_cols:
  result= df[col].max()-df[col].min()
  print(col,":",result)
# 2. varience
for col in numerical_cols:
  print(col,":",df[col].var())

#standard deviation

for col in numerical_cols:
  print(col,":",df[col].std())

# Percentile and quertile
df[numerical_cols].quantile(
    [0.25, 0.50, 0.75, 0.90]
    )

#Interquartile Range (IQR)

for col in numerical_cols:
  Q1=df[col].quantile(0.25)
  Q3=df[col].quantile(0.75)
  IQR=Q3-Q1;
  print(col,":",IQR )
``````

``````python
#                                                     Outlier Detection

# 1. IQR Method
for col in numerical_cols:
  Q1=df[col].quantile(0.25)
  Q3=df[col].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  outliers=df[
      (df[col] < lower) |
        (df[col] > upper)
  ]
  print(f"\n Features:{col}")
  print("Lower Bound:", lower)
  print("Upper Bound:",upper)
  print("Outlier Count:",len(outliers))



# 2. Z-Score Method
from scipy.stats import  zscore
for col in numerical_cols:
  z_scores=zscore(df[col])
  outlier_count=(abs(z_scores)>3).sum()
  print(col,":",outlier_count)
``````

``````python
#                                                                     Distribution Analysis
# in this we will study
    # 1. Normal distribution
    # 2. Right-skewed distribution
    # 3. Left-skewed distribution
    # 4. skewness
    # 5. kurtosis
#                                                     ```Histogram `````
# A histogram devides numerical values into interval calls bins and shows how many observations falls into eachinterval
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8,5))
sns.histplot(
    data=df,
    x='Hours_Studied',
    bins=20,
    kde=True
)

plt.title("Distributions of hours study ")
# plt.xlabel("Hours_Studied")
# plt.ylabel("Number of student ")
plt.show()
``````
