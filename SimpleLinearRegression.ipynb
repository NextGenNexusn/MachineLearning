```python
#Linear Regression is a fundamental supervised machine learning algorithm used to model the
#relactionship between a dependent and one or more independent (input)
#feature by fitting a straight line to the data
```

```python
#step1 import the necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression


# step 2 create dependent and independent variable initially we are using input based feature after we will directly use datasets


#independent variable
x=np.array([
    [1],
    [2],
    [3],
    [4],
    [5]
]
)

#dependent varible
y=np.array([10,20,30,40,50])


#create the model
model=LinearRegression()

#train the mdoel
model.fit(x,y)

# test the model
val=model.predict([[7]])

print(val)

#Drow the digram
```

```python

#Drow the digram
plt.scatter(x,y, color="blue")
plt.xlabel("independent variable")
plt.ylabel("dependent variable")
plt.title("Linear Regression digram")
plt.show()
```
